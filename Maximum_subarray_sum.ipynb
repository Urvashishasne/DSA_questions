### Maximum Subarray Sum 

```python
# Brute Force Solution 
#T.Complexity- O(n3)
def max_sum_subarray_brute(arr):
    n=len(arr)
    maxi=0
    for i in range(n):
        for j in range(i,n):
            sum=0
            for k in range(i,j+1):
                sum+=arr[k]
            maxi=max(maxi,sum)
    return maxi
arr1=[-2,-3,4,-1,-2,1,5,-3]
max_sum_subarray_brute(arr1)
```

```python
#better -O(n2)
def max_sum_subarray_better(arr):
    n=len(arr)
    maxi=0
    for i in range(0,n):
        sum=0
        for j in range(i,n):
            sum+=arr[j]
            maxi=max(maxi,sum)
    return maxi
max_sum_subarray_better(arr1)
```

```python
#optimal - kadanes algorithm -O(n)
def kada_maxsum(arr):
    sum=0
    maxi=arr[0]
    n=len(arr)
    for i in range(n):
        sum+=arr[i]
        if sum<0:
            sum=0
        maxi=max(maxi,sum)
    return maxi
arr1=[-2,-3,4,-1,-2,1,5,-3]
kada_maxsum(arr1)
```
