Finding subarray having Maximum sum 

```python
#O(n^2) 
def maxsum_print(arr):
    n=len(arr)
    maxisum=0
    sindex=0
    eindex=0
    for i in range(0,n):
        sum=0
        for j in range(i,n):
            sum+=arr[j]
            if sum>maxisum:
                sindex=i
                eindex=j
                maxisum=max(maxisum,sum)
    res=[]
    for i in range(sindex,eindex+1):
        res.append(arr[i])
    return res," maximum sum is :", maxisum
arr1=[-2,-3,4,-1,-2,1,5,-3]
maxsum_print(arr1)
```

```python
#optimal - using kadanes algorithm -O(n)
def kada_maxsum(arr):
    sum=0
    maxi=arr[0]
    sindex=0
    eindex=0
    res=[]
    n=len(arr)
    for i in range(n):
        sum+=arr[i]
        if sum<0:
            sum=0
            sindex=i+1
        if sum>maxi:
            eindex=i
        maxi=max(maxi,sum)
    for i in range(sindex,eindex+1):
        res.append(arr[i])
    return maxi,"subarray is : ",res

arr=[-2,-3,-3,-4,-5,4,-1,-2,1,5,-3]
arr1=[-2,-3,4,-1,-2,1,5,-3]
arr2 = [2, 3, -8, 7, -1, 2, 3]
kada_maxsum(arr2)
```
