# spiral matrix 

Input: matrix = [[1,2,3],[4,5,6],[7,8,9]]
Output: [1,2,3,6,9,8,7,4,5]


```python
def spiral_matrix(matrix: list[list[int]]):
    n = len(matrix)
    m = len(matrix[0])
    ans = []
    top, bottom, left, right = 0, n - 1, 0, m - 1
    while top <= bottom and left <= right:
        for i in range(left, right + 1):  # first row: left to right
            ans.append(matrix[top][i])
        top += 1
        for i in range(top, bottom + 1):  # last column: top to bottom
            ans.append(matrix[i][right])
        right -= 1
        if top <= bottom:
            for i in range(right, left - 1, -1):  # bottom row: right to left
                ans.append(matrix[bottom][i])
            bottom -= 1
        if left <= right:
            for i in range(bottom, top - 1, -1):  # first column: bottom to top
                ans.append(matrix[i][left])
            left += 1

    return ans
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
spiral_matrix(matrix)  # This should output: [1, 2, 3, 6, 9, 8, 7, 4, 5]
```
