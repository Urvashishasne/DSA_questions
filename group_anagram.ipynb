```python
#group anagrams- leet code question
from collections import defaultdict
class Solution:
    def groupAnagrams(self, strs: List[str]) -> List[List[str]]:
        anag_map=defaultdict(list)
        result=[]
        for s in strs:
            sorted_s = tuple(sorted(s))
            anag_map[sorted_s].append(s)

        for value in anag_map.values():
            result.append(value)
        return result 
```
