move zeroes to end :
1. store non zeroes to temp variable 
2. 2 pointer approach (optimal)

```python
def zer_rem(arr):
    n=len(arr)
    temp=[]
    #store non zeroes
    for i in range(0,n):
        if arr[i]!=0:
            temp.append(arr[i])
    #move to the front of array 
    for i in range(0,len(temp)):
        arr[i]=temp[i]
    #
    for i in range(len(temp),n):
        arr[i]=0
    return arr
```

```python
#2.
def zer_rem_op(arr):
    j=-1
    # find the position where zero is present in given array (j)
    for i in range(0,len(arr)):
        if arr[i]==0:
            j=i
            break
    for i in range(j+1,len(arr)):
        if arr[i]!=0:
            arr[j],arr[i]=arr[i],arr[j]
            j+1
    return arr
    
```

```python
arr=[2,9,23,23,1,2,56,3,4,3]
zer_rem(arr), zer_rem_op(arr)
```
