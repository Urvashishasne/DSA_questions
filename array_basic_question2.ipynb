```python
#inplace - using set data structure
def remove_dupli(arr):
    s=set()
    n=len(arr)
    for i in range(0,n):
        s.add(arr[i])
    m=len(s)
    for i in range(0,m):
        arr=s
    return arr
arr=[1,2,3,3,3,44,44,5,5,5,6,6,9,4,4,4,4]
remove_dupli(arr)
```

Second Largest Element in an Array without sorting

```python
# first find largest then for second largest check element is not largest element and
# is greater than value of second_largest variable 
def second_largest(arr,n):
    #find largest one
    large=arr[0]
    for i in range(1,n):
        if arr[i]>large:
            large=arr[i]
    print("largest element in array is :",large)
    sec_large=0
    for i in range(n):
        if arr[i]>sec_large and arr[i]!=large:
            sec_large=arr[i]
    return "Second Largest element is :",sec_large

arr=[1,4,5,6,7,8,9,12,45,56,67]
n=len(arr)
second_largest(arr,n)
```

Second largest element without using sorting and in one pass

```python
def sec_lar(arr,n):
    large=arr[0]
    sec =-1
    for i in range(1,n):
        if arr[i]>large:
            sec =large
            large=arr[i]
    return 'second largest element is :',sec
arr=[1,4,5,6,7,8,9,12,45,56,67]
n=len(arr)
sec_lar(arr,n)
```
