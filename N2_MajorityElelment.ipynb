# Majority ELement n/2
A majority element in an array A [] of size n is an element that appears more than n/2 times

Brute force approach 

```python
# Traversing and counting 
def Maj_ele(arr):
    n=len(arr)
    m=n//2
    count=0
    for i in range(0,n):
        arr[i]
        for j in range(n):
            if arr[i]==arr[j]:
                count+=1
        if count>m:
            return arr[i]
    count=0
    return None

#tc=O(n2) sc=O(1)
arr=[2,4,4,3,4,4]
Maj_ele(arr)
```

```python
#use hashing 
def majority_element(arr):
    n=len(arr)
    m=n//2
    hash=[0] * (n)
# hash = [0] * (maxi + 1)
    for i in arr:
        hash[i] +=1
    for i in arr:
        if hash[i] >m:
            return i
    return None
arr=[4,2,4,3,4,4]
majority_element(arr)
#tc - O(2n)
```

Optimal approach :- Moore's voting algorithm 

steps :-
1. declare element = none and count = 0 
2. if count = 0 then el = arr[i] and count=1 
3. elif el == arr[i] count+=1 
4. else count-=1 
5. verify el is the majority element , count of el > n//2 


```python
#Moores voting algo
def majority_elem_moore(arr):
    n=len(arr)
    el=None
    cnt=0
    for i in range(n):
        if cnt == 0:
            cnt = 1
            el = arr[i]
        elif el == arr[i]:
            cnt += 1
        else:
            cnt -= 1
    cnt1 = 0
    for i in range(n):
        if arr[i] == el:
            cnt1 += 1
    if cnt1 > (n // 2):
        return el
    return -1
arr=[4,2,3,4,4,4]
majority_elem_moore(arr)
```
