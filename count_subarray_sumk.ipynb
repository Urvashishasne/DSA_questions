```python
def count_subarrays(arr,k):
    n=len(arr)
    cnt=0
    for i in range(n):
        sm=0
        for j in range(i,n):
            sm+=arr[j]
            if sm==k:
                cnt=cnt+1
                # print the subarray
                for c in range(i,j+1):
                    print(arr[c],end='')
    return cnt
arr=[1,1,2,3,4,5]
count_subarrays(arr,2)
```

mpp = defaultdict(int) creates a dictionary where every new key will have a default value of 0.

preSum keeps track of the cumulative sum (prefix sum) of the elements.

cnt counts the number of subarrays whose sum equals k.

mpp[0] = 1 initializes the map with a prefix sum of 0.

```python
from collections import defaultdict
def findAllSubarraysWithGivenSum(arr, k):
    n = len(arr) # size of the given array.
    mpp = defaultdict(int)
    preSum = 0
    cnt = 0
    mpp[0] = 1 # Setting 0 in the map.
    for i in range(n):
        # add current element to prefix Sum:
        preSum += arr[i]
        # Calculate x-k:
        remove = preSum - k
        # Add the number of subarrays to be removed:
        cnt += mpp[remove]
        # Update the count of prefix sum
        # in the map.
        mpp[preSum] += 1
    return cnt

arr = [3, 1, 2, 4]
k = 6
findAllSubarraysWithGivenSum(arr, k)
    
```
