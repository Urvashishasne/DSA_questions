# Split an array into two equal Sum subarrays

```python
def split_into_subarray(arr):
    n=len(arr)
    for i in range(0,n):
        sum1=0
        sum1+=arr[i]
        sum2=0
        for j in range(i+1,n):
            sum2+=arr[j]
        if sum1==sum2:
            print(arr[:i+1],'and',arr[i+1:])
            return True
    return False
ar=[2,1,1]
split_into_subarray(ar)
```

```python
#second way : sum 
def split2equal_sum(arr):
    total_sum = sum(arr)
    
    # If the total sum is odd, it's impossible to split into two equal subarrays
    if total_sum % 2 != 0:
        return False

    half_sum = total_sum // 2
    running_sum = 0

    for i in range(len(arr)):
        running_sum += arr[i]
        
        if running_sum == half_sum:
            return arr[:i+1], arr[i+1:]
    
    return False

# Example usage
arr = [1,9, 5, 5]
result = split2equal_sum(arr)
if result:
    print("The array can be split into two equal sum subarrays:", result)
else:
    print("The array cannot be split into two equal sum subarrays.")
```
