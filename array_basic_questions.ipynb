```python
#Largest element in array 
#1 define function giving an array and size of it 
def large(arr, n):
    #take first elementof array as maximum 
    max= arr[0]
    #run a loop for a array 
    for i in range(0,n):
        if arr[i]>max:
            max=arr[i]
    return max
arr=[1,443,32,543,63,643,23,54,56,23,6,412,443,63,23,542,5,345,90]
n=len(arr)
large(arr,n)
```

```python
#check if an array is sorted or not 
def issorted(arr):
    n = len(arr)
    for i in range(1, n):
        if arr[i] < arr[i-1]:
            return False
    return True
arr=[1,6,4,6]
issorted(arr)
```

```python
#Linear Search 
def Linearsearch(arr, item):
    n=len(arr)
    for i in range(0,n):
        if (arr[i]==item):
            return i
    return -1

arr=[1,2,3,41,5,6,6,7,32,6,2,1,4,5,2]
Linearsearch(arr,98)
```

left Rotate an array by one place (in place)

```python
def L_rotate_1(arr):
    n= len(arr)
    temp = arr[0]
    for i in range(0,n-1):
        arr[i]=arr[i+1]
    arr[n-1]=temp
    return arr

arr=[1,2,3,4,5]
L_rotate_1(arr)
```

Right rotation byb one place (inplace)

```python
def r_rotation_1(arr):
    size=len(arr)
    temp = arr[size-1]
    for i in range(1,size):# from index 1 to n-1 where size of array is n
        arr[size-i]=arr[size-(i+1)] 
    arr[0]=temp
    return arr
arr=[1,2,3,4,5]
r_rotation_1(arr)
```
