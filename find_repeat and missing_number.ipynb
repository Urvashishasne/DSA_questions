# Find the repeating and missing numbers

Problem Statement: You are given a read-only array of N integers with values also in the range [1, N] both inclusive. Each integer appears exactly once except A which appears twice and B which is missing. The task is to find the repeating and missing numbers A and B where A repeats twice and B is missing.

```python
#brute 
def repeat_miss_num_brute(arr):
    n=len(arr)
    rep_no,mis_no = -1,-1
    for i in range(1,n+1):
        cnt=0
        for j in range(n):
            if arr[j] == i:
                cnt += 1
        if cnt == 0:
            mis_no =i
        elif cnt==2:
            rep_no=i
        if rep_no != -1 and mis_no != -1:
            break
    return ["Missing number is : ",mis_no ,"Repeat number is : ",rep_no]
arr=[3,1,2,5,3]
repeat_miss_num_brute(arr)
```

```python
# better - using hash 
def repeat_miss_num_better(arr):
    n=len(arr)
    repeat=-1;missed=-1
    hash= [0] * (n + 1) # size of hash is n+1(because 0 based indexing) 
    # and everything will be initialize with value 0
    for i in range(n):
        hash[arr[i]] += 1 # num. in ar present i.e. in hash array will be increment with one
        
    for i in range(1, n + 1): # now check counting from 1 to n 
        if hash[i] == 2: # if frequency is 2 that mean number is repeating
            repeat=arr[i]
        elif hash[i] == 0:# if frequency is 0 that mean number is missing
            missed = i
        if repeat != -1 and missed != -1:
            break
    return {repeat ,'missing number :',missed}  
arr=[3,1,2,5,3]
repeat_miss_num_better(arr)
```

```python
# optimal - sum method 
# optimal approach - sum of natural number
def repeat_missed_optimal(arr):
    n=len(arr)
    sumn=(n*(n+1))//2 # sum of n natural number 'sumn', sum of sq. of n ntrl num 'sum2n'
    sum2n= (n * (n + 1) * (2 * n + 1)) // 6
    s, s2 = 0, 0 # sum of ar ele 's' , sum of square of ar ele 's2'
    for i in range(n):
        s+=arr[i]
        s2+=arr[i]*arr[i] 
    val1 = s-sumn  # x - y , repeat - missing no.
    val2 = s2-sum2n # x2 -y2 :- (x-y)(x+y)
    val3=val2//val1
    x=(val3+val1)//2
    y=x-val1
    return [x,y] # [repeat , missing number]
arr=[3,1,2,5,3]
repeat_missed_optimal(arr) 
```
