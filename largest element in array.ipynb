# Largest element in array 

```python
#1 sorting 
def largest1(arr):
    arr.sort()
    return [-1]

arr=[2,3,4,6,3,5,2,3]
largest1(arr)
#Time Complexity: O(N*log(N)) space complexity : o(n)
```

```python
# iterate and compare 
def largest2(arr):
    maxi=arr[0]
    for i in range(len(arr)):
        if arr[i]>maxi:
            maxi=arr[i]
    return maxi
arr=[2,3,4,6,3,5,2,3]
largest2(arr)       
#time complexity : O(n)    Space Complexity: O(1)
```

```python
def largest3(arr):
    n=len(arr)
    maxi=arr[0]
    for i in range(1,n):
        maxi=max(maxi,arr[i])
    return maxi
arr=[2,3,4,6,3,5,2,3]
largest3(arr)
#time complexity : O(n)    Space Complexity: O(1)
```
