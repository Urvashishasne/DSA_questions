find the number that appear once, where other are twice.

```python
#1 linear search for each ele
def oncefinding(nums):
    for i in range(len(nums)):
        no=nums[i]
        cnt=0
        for j in range(len(nums)):
            if nums[j]==no:
                cnt+=1
        if cnt==1:
            return no
#tc= O(n*2)
#sc=O(1)
nums=[1,1,2,3,3,4,4]
oncefinding(nums)
```

```python
#Better Approach(Using Hashing): 
def getSingleElement(arr):
    n = len(arr)
    maxi = max(arr)
    hash = [0] * (maxi + 1)
    for num in arr:
        hash[num] += 1
        print(hash[num])
    for num in arr:
        if hash[num] == 1:
            return num
    return -1

arr = [4, 1, 2, 1, 2]
getSingleElement(arr)
 
```

```python
# using hashmap 
def geteleonce(arr):
    n=len(arr)
    hashmap={}
    for nu in arr:
        hashmap[nu]=hashmap.get(nu,0)+1
    for nu,val in hashmap.items():
        if val==1:
            return nu
    return -1

arr=[1,1,4,4,3,2,3,3]
geteleonce(arr)
```

```python
#using XOR -optimal 

def getSingleElement(arr):
    # XOR all the elements:
    xorr = 0
    for num in arr:
        # xor - a^a =0 -> 0^b=b
        xorr ^=num #(xorr=xorr^num)
    return xorr

arr = [4, 1, 2, 1, 2]
getSingleElement(arr)
```
