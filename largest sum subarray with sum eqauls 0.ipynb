# largest subarray with sum is 0 

Link for this question : https://bit.ly/3w5QSzC

 Given an array containing both positive and negative integers, we have to find the length of the longest subarray with the sum of all elements equal to zero.

```python
def larg_sum_equals_zero(arr):
    n=len(arr)
    ln=0
    malen=0
    for i in range(0,n):
        sum=0
        for j in range(i,n):
            sum+=arr[j]
            if sum==0 :
                ln= len(arr[i:j+1])
            malen=max(ln,malen)
    return malen
```

```python
arr=[9, -3, 3, -1, 6, -5]
larg_sum_equals_zero(arr)
```

```python
# optimal solution- using prefix sum
def larg_sum_equals_zero_opti(arr):
    n=len(arr)
    maxi=0
    sum=0
    hash_ = {}
    for i in range(0,n):
        sum+=arr[i]
        if sum==0:
            maxi=max(maxi,i+1)
        else:
            if sum in hash_ :
                maxi=max(maxi,i-hash_[sum])
            else :
                hash_[sum] = i
    return maxi
```

```python
arr=[9, -3, 3, -1, 6, -5]
larg_sum_equals_zero_opti(arr)
```
