# Leaders in Array

```python
def lead_array_brute(arr):
    n=len(arr)
    ans=[]
    for i in range(0,n):
        leader = True
        for j in range(i+1,n):
            if arr[j]>arr[i]:
                leader = False
                break
        if leader==True:
            ans.append(arr[i])
    return ans 
arr=[10,22,12,3,0,6]
lead_array_brute(arr)  
```

```python
def lead_array_opt(arr):
    n=len(arr)
    lead=[]
    maxi=arr[n-1]
    lead.append(maxi)
    for i in range(n-2,0,-1):
        if arr[i]>maxi:
            lead.append(arr[i])
        maxi=max(arr[i:])
    return lead
arr=[10,22,12,3,0,6]
lead_array_opt(arr)
```

```python
def printLeaders(arr, n):
    ans = []
    max_elem = arr[n - 1]
    ans.append(arr[n - 1])
    for i in range(n - 2, -1, -1):
        if arr[i] > max_elem:
            ans.append(arr[i])
            max_elem = arr[i]

    return ans
printLeaders(arr,6)
```
